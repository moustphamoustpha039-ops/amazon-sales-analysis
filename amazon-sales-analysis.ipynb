```python
import pandas as pd
df = pd.read_csv('/content/amazon_project/amazon.csv')
df.head()
```

```python
# عرض معلومات عن البيانات
print(df.shape)        # عدد الصفوف والأعمدة
print(df.columns)      # أسماء الأعمدة
df.info()              # معلومات تفصيلية
```

```python
df.head()
df.shape
df.info()
df.isnull().sum()  # القيم الناقصة
```

```python
# حذف الصفين الناقصين
df.dropna(subset=['rating_count'], inplace=True)

# تحويل الأسعار من نص إلى أرقام
df['discounted_price'] = df['discounted_price'].str.replace('₹','').str.replace(',','').astype(float)
df['actual_price'] = df['actual_price'].str.replace('₹','').str.replace(',','').astype(float)
df['discount_percentage'] = df['discount_percentage'].str.replace('%','').astype(float)
df['rating'] = pd.to_numeric(df['rating'], errors='coerce')

print("تم التنظيف بنجاح ✅")
df.head()
```

```python
import matplotlib.pyplot as plt

# 1. توزيع التقييمات
plt.figure(figsize=(8,4))
df['rating'].hist(color='steelblue')
plt.title('توزيع تقييمات المنتجات')
plt.xlabel('التقييم')
plt.ylabel('عدد المنتجات')
plt.show()

# 2. أكثر الفئات
plt.figure(figsize=(10,5))
df['category'].str.split('|').str[0].value_counts()[:10].plot(kind='bar', color='coral')
plt.title('أكثر 10 فئات في أمازون')
plt.xticks(rotation=45)
plt.show()

# 3. العلاقة بين الخصم والتقييم
plt.figure(figsize=(8,4))
plt.scatter(df['discount_percentage'], df['rating'], alpha=0.3)
plt.title('الخصم مقابل التقييم')
plt.xlabel('نسبة الخصم')
plt.ylabel('التقييم')
plt.show()
```

```python
import matplotlib.pyplot as plt

# 1. Rating Distribution
plt.figure(figsize=(8,4))
df['rating'].hist(color='steelblue')
plt.title('Product Rating Distribution')
plt.xlabel('Rating')
plt.ylabel('Number of Products')
plt.show()

# 2. Top 10 Categories
plt.figure(figsize=(10,5))
df['category'].str.split('|').str[0].value_counts()[:10].plot(kind='bar', color='coral')
plt.title('Top 10 Categories on Amazon')
plt.xlabel('Category')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# 3. Discount vs Rating
plt.figure(figsize=(8,4))
plt.scatter(df['discount_percentage'], df['rating'], alpha=0.3, color='green')
plt.title('Discount % vs Rating')
plt.xlabel('Discount Percentage')
plt.ylabel('Rating')
plt.show()

# 4. Summary
print("=== Analysis Summary ===")
print(f"Total Products: {len(df)}")
print(f"Average Rating: {df['rating'].mean():.2f}")
print(f"Average Discount: {df['discount_percentage'].mean():.1f}%")
print(f"Max Price: {df['actual_price'].max()}")
print(f"Min Price: {df['actual_price'].min()}")
```

```python
# @title AI prompt cell

import ipywidgets as widgets
from IPython.display import display, HTML, Markdown,clear_output
from google.colab import ai

dropdown = widgets.Dropdown(
    options=[],
    layout={'width': 'auto'}
)

def update_model_list(new_options):
    dropdown.options = new_options
update_model_list(ai.list_models())

text_input = widgets.Textarea(
    placeholder='Ask me anything....',
    layout={'width': 'auto', 'height': '100px'},
)

button = widgets.Button(
    description='Submit Text',
    disabled=False,
    tooltip='Click to submit the text',
    icon='check'
)

output_area = widgets.Output(
     layout={'width': 'auto', 'max_height': '300px','overflow_y': 'scroll'}
)

def on_button_clicked(b):
    with output_area:
        output_area.clear_output(wait=False)
        accumulated_content = ""
        for new_chunk in ai.generate_text(prompt=text_input.value, model_name=dropdown.value, stream=True):
            if new_chunk is None:
                continue
            accumulated_content += new_chunk
            clear_output(wait=True)
            display(Markdown(accumulated_content))

button.on_click(on_button_clicked)
vbox = widgets.GridBox([dropdown, text_input, button, output_area])

display(HTML("""
<style>
.widget-dropdown select {
    font-size: 18px;
    font-family: "Arial", sans-serif;
}
.widget-textarea textarea {
    font-size: 18px;
    font-family: "Arial", sans-serif;
}
</style>
"""))
display(vbox)
```
